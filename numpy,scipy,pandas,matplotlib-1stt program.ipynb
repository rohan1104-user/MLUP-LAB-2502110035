```python
import numpy as np

data = np.array([10, 20, 30, 40])

print(np.mean(data))
print(np.max(data))
```

```python
import pandas as pd

data = {
    "Part": ["P101", "P102", "P103"],
    "Diameter": [25.4, 30.2, 22.5]
}

df = pd.DataFrame(data)

print(df)
print(df["Diameter"].mean())
```

```python
import matplotlib.pyplot as plt

diameters = [25.4, 30.2, 22.5, 27.8]

plt.plot(diameters)
plt.xlabel("Part")
plt.ylabel("Diameter")
plt.title("Part Diameters")
plt.show()
```
